```python
from playwright.sync_api import sync_playwright
import time

URL = "https://www.google.com/maps/search/cafeterias+em+Cotia"

with sync_playwright() as p:

    browser = p.chromium.launch(headless=False)
    context = browser.new_context()
    page = context.new_page()

    page.goto(URL, wait_until="domcontentloaded")

    page.wait_for_timeout(5000)

    cards = page.locator('[role="article"]')

    total = cards.count()

    print(f"{total} cards encontrados.")

    limite = min(total, 10)      # testar apenas os 10 primeiros

    for i in range(limite):

        print(f"\n========== CARD {i+1} ==========")

        # Recria o locator a cada iteração
        # (evita problemas caso o DOM seja atualizado)
        card = page.locator('[role="article"]').nth(i)

        try:

            nome = card.text_content(timeout=3000)

            print(f"Nome:\n{nome[:80]}")

            card.click(timeout=5000)

            page.wait_for_timeout(2500)

            print("Card aberto.")

            # Exemplo: telefone
            telefone = page.locator('button[data-item-id^="phone"]').text_content(timeout=2000)

            print("Telefone:", telefone)

        except Exception as e:

            print("Erro:", e)

        print("Cards restantes:",
              page.locator('[role="article"]').count())

        time.sleep(2)

    browser.close()
```
